```python
import numpy as np
# 1. 五筆輸入資料 X 真實答案 Y
X = np.array([[1,2,3,4,5,6], [2,1,3,5,4,6], [3,2,1,6,5,4], [4,3,2,1,6,5], [5,4,3,2,1,6]], dtype=float)
Y = np.array([[1,0,0], [0,1,0], [0,0,1], [1,0,0], [0,1,0]], dtype=float)
print('X =\n', X, '\nX shape =', X.shape)  # (5, 6)
print('\nY =\n', Y, '\nY shape =', Y.shape)  # (5, 3)

# 2. 權重W與偏值b
W1 = np.array([[0.1,0.2,0.1,0.0,0.1,0.2],[0.2,0.1,0.0,0.1,0.2,0.1],[0.1,0.0,0.2,0.1,0.1,0.2],[0.0,0.1,0.1,0.2,0.2,0.1],[0.2,0.1,0.2,0.1,0.0,0.1]])
b1 = np.array([0.1,0.2,0.1,0.0,0.2])
W2 = np.array([[0.2,0.1,0.3,0.2,0.1],[0.1,0.3,0.2,0.1,0.2],[0.3,0.2,0.1,0.2,0.1],[0.2,0.1,0.2,0.3,0.2]])
b2 = np.array([0.1,0.2,0.1,0.0])
W3 = np.array([[0.2,0.1,0.3,0.2],[0.1,0.3,0.2,0.1],[0.3,0.2,0.1,0.2],[0.2,0.1,0.2,0.3]])
b3 = np.array([0.1,0.0,0.2,0.1])
W4 = np.array([[0.2,0.1,0.3,0.2],[0.1,0.3,0.2,0.1],[0.3,0.2,0.1,0.2]])
b4 = np.array([0.1,0.2,0.0])

# 3. 激發函數。Softmax 需支援批次資料：每一列獨立做正規化。
def relu(z):
    return np.maximum(0, z)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def softmax(z):
    z_stable = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(z_stable)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)

# 4. 批次前向傳播。X shape=(5,6)，權重轉置後與資料相乘。
z1 = X @ W1.T + b1
h1 = relu(z1)
z2 = h1 @ W2.T + b2
h2 = relu(z2)
z3 = h2 @ W3.T + b3
h3 = sigmoid(z3)
z4 = h3 @ W4.T + b4
y_hat = softmax(z4)

# 5. 批次版 Cross Entropy：先算每筆資料的 loss，再取平均。
def cross_entropy_error(y_hat, Y):
    if y_hat.ndim == 1:
        y_hat = y_hat.reshape(1, -1)
        Y = Y.reshape(1, -1)
    batch_size = y_hat.shape[0]
    delta = 1e-7
    return -np.sum(Y * np.log(y_hat + delta)) / batch_size
average_loss = cross_entropy_error(y_hat, Y)

# 6. 列印每一層 W、b、z、h / y_hat 的內容、形狀與參數數量
def show_array(name, value):
    print(f'{name} =\n{value}')
    print(f'{name} shape = {value.shape}\n')
for layer, W, b, z, h in [(1,W1,b1,z1,h1),(2,W2,b2,z2,h2),(3,W3,b3,z3,h3),(4,W4,b4,z4,y_hat)]:
    print(f'========== 第 {layer} 層 ==========')
    show_array(f'W{layer}', W); print(f'W{layer} 參數個數 = {W.size}')
    show_array(f'b{layer}', b); print(f'b{layer} 參數個數 = {b.size}')
    show_array(f'z{layer}', z)
    show_array('y_hat' if layer == 4 else f'h{layer}', h)
print('========== 批次誤差 ==========')
show_array('Y', Y)
print('每筆資料預測類別 =', np.argmax(y_hat, axis=1) + 1)
print('每筆資料 Cross Entropy Error =', -np.sum(Y * np.log(y_hat + 1e-7), axis=1))
print('5 筆資料的平均 Cross Entropy Error =', average_loss)
```
